```python
from google.colab import files
from pathlib import Path
import zipfile
uploaded = files.upload()
name = next(n for n in uploaded if n.endswith(".zip"))
target = Path("/content/etapp2").resolve()
with zipfile.ZipFile(name) as z:
    for member in z.namelist():
        if not (target / member).resolve().is_relative_to(target):
            raise ValueError("Vigane ZIP-faili tee")
    z.extractall(target)
project = next(target.rglob("requirements.txt")).parent
print(project)
```

```python
import subprocess, sys
for args in [["kood/etapp2_demo.py"], ["-m", "unittest", "discover", "-s", "testid", "-v"]]:
    subprocess.run([sys.executable, *args], cwd=project, check=True)
```
