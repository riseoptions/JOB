# Finantsarvestus · etapp 3
Laadi üles Finantsarvestus_etapp3.zip. Käivitus kasutab ainult kaasasolevaid õppeandmeid.

```python
from google.colab import files
from pathlib import Path
import zipfile, subprocess, sys

uploaded = files.upload()
archive = next(n for n in uploaded if n.endswith('.zip'))
dest = Path('/content/finantsarvestus_03'); dest.mkdir(exist_ok=True)
with zipfile.ZipFile(archive) as z:
    for info in z.infolist():
        target=(dest/info.filename).resolve()
        if not target.is_relative_to(dest.resolve()):
            raise ValueError('Lubamatu ZIP tee')
    z.extractall(dest)
root=next(dest.rglob('requirements.txt')).parent
subprocess.run([sys.executable,'-m','pip','install','-r',str(root/'requirements.txt')],check=True)
```

```python
subprocess.run([sys.executable,'kood/etapp3_demo.py'],cwd=root,check=True)
subprocess.run([sys.executable,'-m','unittest','discover','-s','testid','-v'],cwd=root,check=True)
latest=max((root/'valjund/etapp3').glob('*/Finantsanaluus.xlsx'),key=lambda p:p.stat().st_mtime)
files.download(str(latest))
```
